```python
import pandas as pd
```

```python
df_ca = pd.read_csv("data/agg_mean_citytype_CA.csv")
df_fl = pd.read_csv("data/agg_mean_citytype_FL.csv")
df_pa = pd.read_csv("data/agg_mean_citytype_PA.csv")

agg_mean = pd.concat([df_ca, df_fl, df_pa], ignore_index=True)
agg_mean.rename(columns={"time_period":"date"},inplace=True)
agg_mean["is_prediction"] = False


agg_mean_doc = agg_mean[["business_id", "date", "doc_sentiment","category_avg_by_city","is_prediction"]].copy()
#agg_mean_city = agg_mean[["business_id", "date", "category_avg_by_city","is_prediction"]].copy()
```

```python
agg_mean_doc.info()
```

```python
agg_mean_doc.to_csv("data/agg_mean_doc.csv", index=False)
```

```python
df_ca = pd.read_csv("data/CA_prediction_WithName.csv")
df_fl = pd.read_csv("data/FL_prediction_WithName.csv")
df_pa = pd.read_csv("data/PA_prediction_WithName.csv")
predicted_vadar = pd.concat([df_ca, df_fl, df_pa], ignore_index=True)
```

```python
predicted_vadar.rename(columns={"Business_ID": "business_id", "Date": "date", "Predicted_VADAR": "doc_sentiment"}, inplace=True)
predicted_vadar.drop(["Unnamed: 0", "name"], axis=1, inplace=True)
predicted_vadar["category_avg_by_city"] = None
predicted_vadar["is_prediction"] = True
predicted_vadar.info()
```

```python
predicted_vadar.to_csv("data/predicted_vadar.csv", index=False)
```

```python
time_series = pd.concat([agg_mean_doc, predicted_vadar])
time_series.sort_values(by="business_id", inplace=True)
```

```python
time_series.to_csv("data/time_series.csv", index=False)
```
