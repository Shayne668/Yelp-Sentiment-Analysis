```python
import csv
import pandas as pd
```

```python
business = pd.read_json('../yelp_data/yelp_academic_dataset_business.json', lines=True)
```

```python
business
```

```python
split_categories = business['categories'].str.split(',', expand = True)
split_categories
```

```python
ids = business[['business_id']]
ids
```

```python
stacked = pd.concat([ids, split_categories], axis = 1).set_index('business_id').stack()
stacked
```

```python
id_and_category = stacked.reset_index().rename(columns={0: 'category'}).drop('level_1', axis = 1)
id_and_category
```

```python
id_and_name = business[['business_id', 'name']]
merged = id_and_category.merge(id_and_name, on = 'business_id')
id_name_category = merged[['business_id', 'name', 'category']]
id_name_category
```

```python
# Write to csv
id_name_category.to_csv('../Data/id_name_category.csv', index = False)
```
